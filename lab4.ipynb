```python

from collections import deque

# Initial state: all disks are on rod A
# State represents the disks on rods A, B, C
start = ((3, 2, 1), (), ())
goal = ((), (), (3, 2, 1))

def successors(state):
    result = []

    for i in range(3):
        if not state[i]:
            continue

        disk = state[i][-1]

        for j in range(3):
            if i == j:
                continue

            # Move only if destination is empty
            # or its top disk is larger
            if not state[j] or state[j][-1] > disk:
                new_state = [list(rod) for rod in state]

                new_state[i].pop()
                new_state[j].append(disk)

                result.append(tuple(tuple(rod) for rod in new_state))

    return result


def bfs(start, goal):
    queue = deque([(start, [start])])
    visited = {start}

    while queue:
        state, path = queue.popleft()

        if state == goal:
            return path

        for next_state in successors(state):
            if next_state not in visited:
                visited.add(next_state)
                queue.append((next_state, path + [next_state]))

    return None


# Run the search
path = bfs(start, goal)

if path:
    print("Solution found!")
    print("Minimum moves:", len(path) - 1)

    for step, state in enumerate(path):
        print(f"\nStep {step}:")
        print("A:", state[0])
        print("B:", state[1])
        print("C:", state[2])
else:
    print("No solution found.")
```
