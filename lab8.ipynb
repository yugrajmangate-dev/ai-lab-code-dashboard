```python
import re
import random

def chatbot(user_input):
    text = user_input.lower().strip()

    if re.search(r"\b(hello|hi|hey)\b", text):
        return "Hello! How can I help you?"

    elif re.search(r"\bhow are you\b", text):
        return "I am fine! Thanks for asking."

    elif re.search(r"\b(your name|who are you)\b", text):
        return "I am a simple rule-based chatbot."

    elif re.search(r"\b(thank you|thanks)\b", text):
        return "You're welcome!"

    elif re.search(r"\b(bye|goodbye|exit|quit)\b", text):
        return "Goodbye! Have a nice day."

    else:
        return "Sorry, I don't understand."

print("Chatbot: Hello! Welcome to my chatbot.")
print("Chatbot: Type 'bye' to exit.")

while True:
    user_input = input("You: ")
    print("User Input:", user_input)

    response = chatbot(user_input)
    print("Chatbot Response:", response)

    if re.search(r"\b(bye|goodbye|exit|quit)\b", user_input.lower()):
        break
```
