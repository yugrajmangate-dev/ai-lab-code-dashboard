Use BFS to solve a planning problem 

examples:
1.  8-Puzzle Problem 
2.Robot Path planning Problem 
3.Blocks World

**8-Puzzle**


```python

from collections import deque

def bfs(start, goal):
    start = tuple(start)
    goal = tuple(goal)

    queue = deque([(start, [start])])
    visited = {start}

    while queue:
        state, path = queue.popleft()

        if state == goal:
            return path

        zero = state.index(0)
        row, col = divmod(zero, 3)

        moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

        for dr, dc in moves:
            r, c = row + dr, col + dc

            if 0 <= r < 3 and 0 <= c < 3:
                new_zero = r * 3 + c
                new_state = list(state)

                new_state[zero], new_state[new_zero] = (
                    new_state[new_zero], new_state[zero]
                )

                new_state = tuple(new_state)

                if new_state not in visited:
                    visited.add(new_state)
                    queue.append((new_state, path + [new_state]))

    return None


start = [
    0, 1, 3,
    4, 2, 6,
    7, 5, 8
]

goal = [
    1, 2, 3,
    4, 5, 6,
    7, 8, 0
]

result = bfs(start, goal)

if result is not None:
    print("Solution found!\n")

    for step, state in enumerate(result):
        print(f"Step {step}:")
        for i in range(0, 9, 3):
            print(state[i:i + 3])
        print()

    print("Total moves:", len(result) - 1)

else:
    print("No solution found.")
```

```python
from collections import deque

def bfs(grid, start, goal):
    rows, cols = len(grid), len(grid[0])
    queue = deque([(start, [start])])
    visited = {start}

    moves = [
        (-1, 0), (1, 0),
        (0, -1), (0, 1)
    ]

    while queue:
        (r, c), path = queue.popleft()

        if (r, c) == goal:
            return path

        for dr, dc in moves:
            nr, nc = r + dr, c + dc

            if (0 <= nr < rows and
                0 <= nc < cols and
                grid[nr][nc] == 0 and
                (nr, nc) not in visited):

                visited.add((nr, nc))
                queue.append(((nr, nc), path + [(nr, nc)]))

    return None


# 0 = Free cell, 1 = Obstacle
grid = [
    [0, 0, 0, 0],
    [1, 1, 0, 1],
    [0, 0, 0, 0],
    [0, 1, 1, 0]
]

start = (0, 0)
goal = (3, 3)

path = bfs(grid, start, goal)

if path:
    print("Shortest path:", path)
    print("Number of moves:", len(path) - 1)

    for r, c in path:
        if (r, c) not in (start, goal):
            grid[r][c] = 2

    print("\nGrid with path:")
    for row in grid:
        print(row)
else:
    print("No path found")
```

```python
from collections import deque

def bfs(start, goal):
    start = tuple(tuple(col) for col in start)
    goal = tuple(tuple(col) for col in goal)

    queue = deque([start])
    parent = {start: None}
    action = {}

    while queue:
        state = queue.popleft()

        if state == goal:
            plan = []

            while parent[state] is not None:
                plan.append((action[state], state))
                state = parent[state]

            plan.reverse()
            return plan

        for i in range(3):
            if not state[i]:
                continue

            block = state[i][-1]

            for j in range(3):
                if i == j:
                    continue

                new_state = [list(col) for col in state]
                new_state[i].pop()
                new_state[j].append(block)

                new_state = tuple(tuple(col) for col in new_state)

                if new_state not in parent:
                    parent[new_state] = state
                    action[new_state] = (
                        f"Move block {block} from column {i+1} "
                        f"to column {j+1}"
                    )
                    queue.append(new_state)

    return None


start = [
    ["A", "B"],
    ["C"],
    []
]

goal = [
    [],
    [],
    ["A", "B", "C"]
]

plan = bfs(start, goal)

print("Initial State:", start)

if plan is not None:
    for step, (move, state) in enumerate(plan, 1):
        print(f"\nStep {step}: {move}")

        for i, col in enumerate(state, 1):
            print(f"Column {i}: {list(col)}")

    print("\nGoal State Reached!")
    print("Total Moves:", len(plan))

else:
    print("No solution found")
```
