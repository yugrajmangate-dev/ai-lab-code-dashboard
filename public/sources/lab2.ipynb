```python
from collections import deque

# -----------------------------------------
# GRID ENVIRONMENT
# -----------------------------------------
grid = [
    ['S', '.', '.', '#', '.'],
    ['#', '#', '.', '#', '.'],
    ['.', '.', '.', '.', '.'],
    ['.', '#', '#', '#', '.'],
    ['.', '.', '.', '.', 'G']
]

ROWS = len(grid)
COLS = len(grid[0])

start = (0, 0)
goal = (4, 4)


# -----------------------------------------
# MODEL-BASED INTELLIGENT AGENT
# -----------------------------------------
class ModelBasedAgent:

    def __init__(self, start, goal):
        self.position = start
        self.goal = goal

        # Internal state representation
        self.internal_map = {}

        # Cells already visited
        self.visited = set()

        # Path followed by the agent
        self.path = [start]

    # -------------------------------------
    # PERCEIVE ENVIRONMENT
    # -------------------------------------
    def perceive(self, grid):
        r, c = self.position

        # Observe current position
        self.internal_map[(r, c)] = grid[r][c]

        # Observe neighboring cells
        directions = {
            "UP": (-1, 0),
            "DOWN": (1, 0),
            "LEFT": (0, -1),
            "RIGHT": (0, 1)
        }

        for action, (dr, dc) in directions.items():
            nr = r + dr
            nc = c + dc

            if 0 <= nr < ROWS and 0 <= nc < COLS:
                self.internal_map[(nr, nc)] = grid[nr][nc]

    # -------------------------------------
    # CHECK WHETHER CELL IS VALID
    # -------------------------------------
    def is_valid(self, position):
        r, c = position

        if r < 0 or r >= ROWS:
            return False

        if c < 0 or c >= COLS:
            return False

        # Avoid known obstacles
        if self.internal_map.get(position) == '#':
            return False

        return True

    # -------------------------------------
    # FIND PATH USING INTERNAL MODEL
    # -------------------------------------
    def find_path(self):
        queue = deque()

        queue.append((self.position, [self.position]))

        visited = set()
        visited.add(self.position)

        directions = [
            (-1, 0),  # UP
            (1, 0),   # DOWN
            (0, -1),  # LEFT
            (0, 1)    # RIGHT
        ]

        while queue:
            current, path = queue.popleft()

            if current == self.goal:
                return path

            r, c = current

            for dr, dc in directions:
                next_position = (r + dr, c + dc)

                if (next_position not in visited and self.is_valid(next_position) ):
                    
                    visited.add(next_position)

                    new_path = path + [next_position]

                    queue.append(
                        (next_position, new_path)
                    )

        return None

    # -------------------------------------
    # SELECT ACTION
    # -------------------------------------
    def select_action(self, next_position):
        r, c = self.position
        nr, nc = next_position

        if nr == r - 1:
            return "UP"

        elif nr == r + 1:
            return "DOWN"

        elif nc == c - 1:
            return "LEFT"

        elif nc == c + 1:
            return "RIGHT"

        return "STOP"

    # -------------------------------------
    # MOVE AGENT
    # -------------------------------------
    def move(self, next_position):
        action = self.select_action(next_position)

        print("Current Position :", self.position)
        print("Action :", action)

        self.position = next_position

        self.visited.add(self.position)
        self.path.append(self.position)

        print("New Position :", self.position)
        print("--------------------------------")

    # -------------------------------------
    # RUN AGENT
    # -------------------------------------
    def run(self, grid):
        print("\n===== MODEL-BASED AGENT =====\n")

        self.visited.add(self.position)

        while self.position != self.goal:

            # Step 1: Perceive environment
            self.perceive(grid)

            # Step 2: Update internal model
            print("Internal State:")
            print(self.internal_map)

            # Step 3: Find path using model
            path = self.find_path()

            if path is None:
                print("No path to goal found.")
                return

            # Get next position
            if len(path) > 1:
                next_position = path[1]

                # Step 4: Select action and move
                self.move(next_position)

        print("\nGoal Reached!")
        print("Goal Position :", self.goal)
        print("Complete Path :", self.path)
        print("Visited Cells :", self.visited)


# -----------------------------------------
# DISPLAY GRID
# -----------------------------------------
def display_grid(grid):
    print("\nGrid Environment:\n")

    for row in grid:
        print(" ".join(row))


# -----------------------------------------
# MAIN PROGRAM
# -----------------------------------------
display_grid(grid)

agent = ModelBasedAgent(start, goal)
agent.run(grid)
```

```python

from collections import deque

grid = [
    ['S', '.', '.', '#', '.'],
    ['#', '#', '.', '#', '.'],
    ['.', '.', '.', '.', '.'],
    ['.', '#', '#', '#', '.'],
    ['.', '.', '.', '.', 'G']
]

ROWS = len(grid)
COLS = len(grid[0])

start = (0, 0)
goal = (4, 4)

directions = [(-1, 0), (1, 0), (0, -1), (0, 1)]

def is_valid(position):
    r, c = position

    if r < 0 or r >= ROWS or c < 0 or c >= COLS:
        return False

    if grid[r][c] == '#':
        return False

    return True

queue = deque([(start, [start])])
visited = {start}

while queue:
    position, path = queue.popleft()

    if position == goal:
        break

    r, c = position

    for dr, dc in directions:
        next_pos = (r + dr, c + dc)

        if is_valid(next_pos) and next_pos not in visited:
            visited.add(next_pos)
            queue.append((next_pos, path + [next_pos]))

else:
    path = []

print("Path:", path)
print("Path length:", len(path) - 1 if path else 0)
```
