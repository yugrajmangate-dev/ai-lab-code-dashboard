**Missionaries and Cannibals**

Question: In the missionaries and cannibals problem, three missionaries and three cannibals must cross a river using a boat which can carry at most two people, under the constraint that, for both banks, if there are missionaries present on the bank, they cannot be outnumbered by cannibals (if they were, the cannibals would eat the missionaries). The boat cannot cross the river by itself with no people on board.

Note : Missionaries == Man , cannibals == ghost

1.Water jug problem


```python
def dfs(state, path, visited):
    x, y = state

    if x == 2:
        return path

    visited.add(state)

    successors = [
        (4, y),                  # Fill 4-litre jug
        (x, 3),                  # Fill 3-litre jug
        (0, y),                  # Empty 4-litre jug
        (x, 0),                  # Empty 3-litre jug
        (x - min(x, 3-y), y + min(x, 3-y)),  # Pour 4 to 3
        (x + min(y, 4-x), y - min(y, 4-x))   # Pour 3 to 4
    ]

    for next_state in successors:
        if next_state not in visited:
            result = dfs(next_state, path + [next_state], visited)
            if result:
                return result

    return None


start = (0, 0)
result = dfs(start, [start], set())

if result:
    print("Solution Path:")
    for state in result:
        print(state)
    print("Total moves:", len(result) - 1)
else:
    print("No solution found")
```

```python
def safe(m, c):
    if not (0 <= m <= 3 and 0 <= c <= 3):
        return False

    mr, cr = 3 - m, 3 - c

    return ((m == 0 or m >= c) and
            (mr == 0 or mr >= cr))


def dfs(state, path, visited):
    m, c, boat = state

    if state == (0, 0, 1):
        return path

    visited.add(state)

    moves = [(1, 0), (2, 0), (0, 1), (0, 2), (1, 1)]

    for dm, dc in moves:
        if boat == 0:
            new_state = (m - dm, c - dc, 1)
        else:
            new_state = (m + dm, c + dc, 0)

        if (safe(new_state[0], new_state[1])
                and new_state not in visited):
            result = dfs(new_state, path + [new_state], visited)

            if result:
                return result

    return None


start = (3, 3, 0)
result = dfs(start, [start], set())

if result:
    print("Solution Path:")
    for state in result:
        print(state)
    print("Total crossings:", len(result) - 1)
else:
    print("No solution found")
```

```python
def safe(state):
    f, w, g, c = state

    if w == g and f != w:
        return False

    if g == c and f != g:
        return False

    return True


def dfs(state, path, visited):
    if state == (1, 1, 1, 1):
        return path

    visited.add(state)

    f, w, g, c = state

    # Farmer crosses alone or takes one item
    moves = [None, 1, 2, 3]

    for item in moves:
        new_state = list(state)
        new_state[0] = 1 - f

        if item is not None:
            if state[item] != f:
                continue
            new_state[item] = 1 - state[item]

        new_state = tuple(new_state)

        if safe(new_state) and new_state not in visited:
            result = dfs(new_state, path + [new_state], visited)

            if result:
                return result

    return None


start = (0, 0, 0, 0)
result = dfs(start, [start], set())

if result:
    names = ["Farmer", "Wolf", "Goat", "Cabbage"]

    print("Solution Path:")
    for state in result:
        print(state)

    print("Total crossings:", len(result) - 1)
else:
    print("No solution found")
```

```python
def dfs(state, goal, path, visited):
    if state == goal:
        return path

    visited.add(state)

    for i in range(3):
        if not state[i]:
            continue

        block = state[i][-1]

        for j in range(3):
            if i == j:
                continue

            new_state = [list(col) for col in state]

            new_state[i].pop()
            new_state[j].append(block)

            new_state = tuple(tuple(col) for col in new_state)

            if new_state not in visited:
                result = dfs(
                    new_state, goal, path + [new_state], visited
                )

                if result:
                    return result

    return None


start = (
    ("A", "B"),
    ("C",),
    ()
)

goal = (
    ("A",),
    ("B",),
    ("C",)
)

result = dfs(start, goal, [], set())

if result is not None:
    print("Initial State:", start)

    for step, state in enumerate(result, 1):
        print(f"\nStep {step}:")
        for i, col in enumerate(state, 1):
            print(f"Column {i}: {list(col)}")

    print("\nGoal State Reached!")
    print("Total moves:", len(result))
else:
    print("No solution found")
```
